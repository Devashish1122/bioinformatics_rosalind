####**TASK BP21.2** <font color=orange> **Complete Rosalind #8  Translating RNA into Protein**</font>

````python
codon_table = {
    "UUU":"F", "UUC":"F", "UUA":"L", "UUG":"L",
    "UCU":"S", "UCC":"S", "UCA":"S", "UCG":"S",
    "UAU":"Y", "UAC":"Y", "UAA":"STOP", "UAG":"STOP",
    "UGU":"C", "UGC":"C", "UGA":"STOP", "UGG":"W",

    "CUU":"L", "CUC":"L", "CUA":"L", "CUG":"L",
    "CCU":"P", "CCC":"P", "CCA":"P", "CCG":"P",
    "CAU":"H", "CAC":"H", "CAA":"Q", "CAG":"Q",
    "CGU":"R", "CGC":"R", "CGA":"R", "CGG":"R",

    "AUU":"I", "AUC":"I", "AUA":"I", "AUG":"M",
    "ACU":"T", "ACC":"T", "ACA":"T", "ACG":"T",
    "AAU":"N", "AAC":"N", "AAA":"K", "AAG":"K",
    "AGU":"S", "AGC":"S", "AGA":"R", "AGG":"R",

    "GUU":"V", "GUC":"V", "GUA":"V", "GUG":"V",
    "GCU":"A", "GCC":"A", "GCA":"A", "GCG":"A",
    "GAU":"D", "GAC":"D", "GAA":"E", "GAG":"E",
    "GGU":"G", "GGC":"G", "GGA":"G", "GGG":"G"
}

with open("/content/rosalind_prot.txt", "r") as file:
  rna = file.read().strip()

protein = ""

for i in range(0, len(rna), 3):
  codon = rna[i:i+3]
  amino_acid = codon_table[codon]

  if amino_acid == "STOP":
    break

  protein += amino_acid

print(protein)
````

#<font color=blue>**P23.**</font> SeqUtils.gc_fraction( ) – a function to compute GC content

####**TASK P23.1** Modify the script from TASK P22.1 so that it prints the GC content for each sequence in the file

####**TASK P23.2** <font color=orange> **Complete Rosalind #9  Computing GC Content**</font>

````python
with open("/content/rosalind_gc.txt", "r") as file:
  lines = file.readlines()

sequences = {}
current_id = ""

for line in lines:
  line = line.strip()

  if line.startswith(">"):
    current_id = line[1:]
    sequences[current_id] = ""
  else:
    sequences[current_id] += line

highest_id = ""
highest_gc = 0

for dna_id, sequence in sequences.items():
  gc_content = (sequence.count("G") + sequence.count("C")) / len(sequence) * 100

  if gc_content > highest_gc:
    highest_gc = gc_content
    highest_id = dna_id

print(highest_id)
print(highest_gc)
````
